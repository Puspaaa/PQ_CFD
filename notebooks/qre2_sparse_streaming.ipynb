# QRE2 Sparse Streaming Check

This sandbox looks at one small classical object: the periodic D2Q9
streaming matrix for the shifted QRE2 validation path.

The point is not to estimate quantum resources. The point is to make
the operator concrete before it becomes a resource-bookkeeping box.

```python
spatial_shape = (2, 3)
```

```python
streaming = qre2_sparse.build_sparse_streaming_matrix(spatial_shape)
```

```python
column_sums = np.asarray(streaming.sum(axis=0)).reshape(-1)
row_sums = np.asarray(streaming.sum(axis=1)).reshape(-1)
is_permutation = bool(
    np.allclose(column_sums, 1.0) and np.allclose(row_sums, 1.0)
)
density = streaming.nnz / (streaming.shape[0] * streaming.shape[1])

display(Markdown(
    f"""
    ## Falsifiable Check

    For a periodic streaming operator, every source population should move
    to exactly one destination, and every destination population should
    receive exactly one source.

    - Spatial grid: `{spatial_shape}`
    - Matrix shape: `{streaming.shape}`
    - Nonzero entries: `{streaming.nnz}`
    - Matrix density: `{density:.4f}`
    - Permutation check passed: `{is_permutation}`
    """
))
```

```python
fig, ax = plt.subplots(figsize=(5.5, 5.5))
ax.spy(streaming, markersize=5)
ax.set_title("Sparse periodic D2Q9 streaming permutation")
ax.set_xlabel("source population index")
ax.set_ylabel("destination population index")
fig.tight_layout()

display(Markdown(
    "## Visual Object\n\n"
    "Each dot is a nonzero entry in the sparse streaming matrix."
))
display(fig)
```
